```python
grocery_list = []
print("Empty grocery list:", grocery_list)
```

```python
grocery_list = ["Apples", "Bread", "Milk", "Cheese", "Eggs"]
print("Grocery list:", grocery_list)
print("Number of items:", len(grocery_list))
```

```python
grocery_list.append("Butter")
print("Grocery list after adding Butter:", grocery_list)
```

```python
print("Last item:", grocery_list[-1])
print("Entire grocery list:")
print(grocery_list)
```

```python
print("Important items:")
print(grocery_list[1])
print(grocery_list[2])
```

```python
grocery_list[3] = "Yogurt"
print("Grocery list after replacing Cheese with Yogurt:")
print(grocery_list)
```

```python
grocery_list.remove("Apples")
print("Final grocery list:")
print(grocery_list)
```
