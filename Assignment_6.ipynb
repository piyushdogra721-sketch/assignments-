```python
#Problem 1: BUGGY CODE
#fix the above problem with super in python
class Person:

  def __init__(self, name, id):
    self.name = name
    self.id = id

  def Display(self):
    print("Name:", self.name)
    print("ID:", self.id)


class Emp(Person):
  def __init__(self, name, id):
    super().__init__(name, id)

  def Print(self):
    print("Employee class called")

Emp_details = Emp("Mayank", 103)

Emp_details.Display()

Emp_details.Print()
```

```python
#problem 2 : Write a python program to find anagram check
def anagram_check(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    if sorted(str1) == sorted(str2):
        return True
    else:
        return False

str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

if anagram_check(str1, str2):
    print("The strings are Anagrams.")
else:
    print("The strings are not Anagrams.")
```

```python
#Problem 3: Write a python program to check for armstrong number
num = int(input("Enter a number: "))
original_num = num

digits = len(str(num))
sum = 0
while num > 0:
    digit = num % 10
    sum = sum + digit ** digits
    num = num // 10

if sum == original_num:
    print(original_num, "is an Armstrong Number.")
else:
    print(original_num, "is not an Armstrong Number.")
```

```python
#Problem 4: Deal with classes and objects
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "White"
        self.seating_capacity = 0

    def assign_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)

car1 = Vehicle(200, 20)

car1.assign_seating_capacity(5)

car2 = Vehicle(180, 25)

car2.assign_seating_capacity(4)

print("----- Car 1 Details -----")
car1.display_properties()

print()

print("----- Car 2 Details -----")
car2.display_properties()
```
